```python
for size in [1, 2]:
    rng = random.Random(42)
    estimates = []
    for _ in range(8):
        selected = rng.choices(data, k=size)
        estimates.append(sum(-2*y*x for x, y in selected)/size)
    print("batch size:", size, "gradients at w=0:", estimates)
```
