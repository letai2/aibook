<div dir="rtl" style="text-align:right">
<p style="text-align:right">پیش از اجرا، ترتیب احتمال‌های سه امتیاز ۲،۱،۰ را حدس بزنید. سپس بپرسید: اگر به همهٔ امتیازها ۱۰۰۰ اضافه کنیم، آیا توزیع باید عوض شود؟</p>
</div>

```python
import math
import matplotlib.pyplot as plt
def softmax(logits):
    maximum = max(logits)
    values = [math.exp(z - maximum) for z in logits]
    return [value / sum(values) for value in values]

def loss(logits, target):
    if type(target) is not int or not 0 <= target < len(logits):
        raise ValueError("Target must be a valid integer class index")
    maximum = max(logits)
    return maximum - logits[target] + math.log(sum(math.exp(z-maximum) for z in logits))

logits = [2.0, 1.0, 0.0]
probabilities = softmax(logits)
print(probabilities, "sum:", sum(probabilities))
print("Target 0 Loss:", loss(logits, 0), "Target 2 Loss:", loss(logits, 2))
assert math.isclose(loss(logits, 0), 0.4076059644)
assert all(math.isclose(a,b) for a,b in zip(probabilities, softmax([1002,1001,1000])))
```

<div dir="rtl" style="text-align:right">
<h2>فقط یک امتیاز را تغییر دهید</h2><p style="text-align:right">امتیاز Class دوم را از ۱ به ۳ ببرید. قبل از اجرا حدس بزنید احتمال Class اول و Loss با هدف صفر چه تغییری می‌کند. شمارهٔ Class فقط شناسه است، نه مقدار کمّیِ پاسخ.</p>
</div>

```python
modified = [2.0, 3.0, 0.0]
print("Modified probabilities:", softmax(modified))
print("Modified target-0 Loss:", loss(modified, 0))
fig, ax = plt.subplots(figsize=(6, 3))
ax.bar([i-0.18 for i in range(3)], probabilities, width=0.36, label="Original")
ax.bar([i+0.18 for i in range(3)], softmax(modified), width=0.36, label="Change one logit")
ax.set(xlabel="Class ID", ylabel="Probability", xticks=range(3), ylim=(0,1))
ax.legend()
plt.show()
assert softmax(modified)[0] < probabilities[0]
```

<div dir="rtl" style="text-align:right">
<p style="text-align:right">در محاسبهٔ مستقیم exp، حتی قبل از تقسیم ممکن است عدد از دامنه بیرون برود. خطاهای زیر عمدی‌اند؛ روش پایدار اختلاف امتیازها را نگه می‌دارد. هدف منفی را هم صریحاً رد می‌کنیم تا index منفیِ Python معنای مسئله را عوض نکند.</p>
</div>

```python
try:
    math.exp(1002)
except OverflowError as error:
    print("Expected overflow:", error)
else:
    raise AssertionError("Expected overflow")
try:
    loss(logits, -1)
except ValueError as error:
    print("Expected target error:", error)
else:
    raise AssertionError("Expected invalid target")
```
