<h2>فقط یک عامل را تغییر دهید</h2>
<p>فقط Learning Rate را برای تابع (w-2)² تغییر دهید. همهٔ مسیرها از w=-3 شروع می‌شوند؛ این تابع با مثال اصلی متفاوت است و نرخ‌هایش نسخهٔ عمومی نیستند.</p>
</div>

```python
for rate in [0.1, 1., 1.1]:
    value = -3.
    path = [value]
    for _ in range(6):
        value -= rate*2*(value-2)
        path.append(value)
    print(rate, path)
```
