<div dir="rtl" style="text-align:right">
<h2 style="text-align:right">فقط یک عامل را تغییر دهید</h2>
<p style="text-align:right">فقط محل نگهداری <bdi dir="ltr">Vocabulary</bdi> را تغییر دهید: از حافظه به فایل موقت و دوباره به حافظه. ترتیب و خروجی باید ثابت بمانند؛ فایل موقت پس از آزمایش حذف می‌شود.</p>
</div>

```python
with TemporaryDirectory() as directory:
    path = Path(directory)/"vocabulary.json"
    tokenizer.save(path)
    restored = CharacterTokenizer.load(path)
    print("Preserved order:", restored.id_to_token == vocabulary)
    assert restored.encode(text) == tokenizer.encode(text)
print("Unknown round trip:", tokenizer.decode(tokenizer.encode("ژچ")))
```
