<p style="text-align:right">فقط فاصلهٔ دو امتیاز را تغییر دهید: <code dir="ltr" style="direction:ltr;text-align:left;unicode-bidi:isolate">[0,gap]</code> برای <code dir="ltr" style="direction:ltr;text-align:left;unicode-bidi:isolate">gap</code>های صفر، یک و دو. ترتیب ثابت است، اما اطمینان چطور تغییر می‌کند؟</p>

```python
for gap in [0, 1, 2]:
    weights = [math.exp(-gap), 1.]
    print("gap:", gap, "probabilities:", [w/sum(weights) for w in weights])
```
