<div dir="rtl" style="text-align:right">
<h2 style="text-align:right">فقط یک عامل را تغییر دهید</h2>
<p style="text-align:right">از <bdi dir="ltr">Layer Value</bdi> یک کپی مستقل بگیرید و فقط وزن آن را دو برابر کنید؛ <bdi dir="ltr">Bias</bdi> ثابت بماند. اینجا خروجی خود <bdi dir="ltr">Linear</bdi> را مشاهده می‌کنیم. <code dir="ltr" style="direction:ltr;text-align:left;unicode-bidi:isolate">Q</code> و <code dir="ltr" style="direction:ltr;text-align:left;unicode-bidi:isolate">K</code> ثابت می‌مانند؛ چرا خروجی <code dir="ltr" style="direction:ltr;text-align:left;unicode-bidi:isolate">V</code> با <bdi dir="ltr">Bias</bdi> غیرصفر الزاماً دو برابر نمی‌شود؟</p>
</div>

```python
from copy import deepcopy
changed_value = deepcopy(value)
with torch.no_grad():
    q_before,k_before = query(x),key(x)
    old_v = value(x)
    changed_value.weight.mul_(2)
    changed_v = changed_value(x)
    print('Q/K unchanged:',torch.equal(q_before,query(x)),torch.equal(k_before,key(x)))
    print('V change:',changed_v-old_v)
    print('difference from twice the old V:',changed_v-2*old_v)
```

<h2 style="text-align:right">خرابی را پیدا کنید</h2>
<p style="text-align:right">دو <bdi dir="ltr">Projection</bdi> زیر <bdi dir="ltr">Weight</bdi> یکسان ولی <bdi dir="ltr">Bias</bdi> متفاوت دارند. تابع <code dir="ltr" style="direction:ltr;text-align:left;unicode-bidi:isolate">copy_projection(source, destination)</code> را اصلاح کنید تا هر دو را داخل <code dir="ltr" style="direction:ltr;text-align:left;unicode-bidi:isolate">no_grad</code> کپی کند و <code dir="ltr" style="direction:ltr;text-align:left;unicode-bidi:isolate">destination</code> را برگرداند.</p>
</div>

```python
source,destination = nn.Linear(3,2),nn.Linear(3,2)
with torch.no_grad():
    source.bias.fill_(1.)
    destination.bias.zero_()
    destination.weight.copy_(source.weight)
print('weight-only copy difference:',(source(x)-destination(x)).detach())
```
