<div dir="rtl">
<h2>فقط یک عامل را تغییر دهید</h2>
<p>فقط فراوانی «گرم» را از یک به ده تغییر دهید. شمارش جفت‌ها را تازه بسازید و ببینید نخستین انتخاب چه تغییری می‌کند؛ قانون تساوی ثابت است.</p>
</div>

```python
for frequency in [1, 10]:
    counts = Counter()
    for word, count in [("کار", 2), ("کارگر", 1), ("گرم", frequency)]:
        for pair in zip(word, word[1:]):
            counts[pair] += count
    chosen = min(counts, key=lambda pair: (-counts[pair], pair))
    print("Frequency:", frequency, "Chosen pair:", chosen, "Counts:", counts)
```

<div dir="rtl">
<h2>خرابی را پیدا کنید</h2>
<p>قواعد ذخیره‌شده باید به ترتیب آموخته‌شده اجرا شوند. تابع apply_rules(tokens,rules) تمام قواعد را با ادغام غیرهم‌پوشان اجرا کند؛ مستقل از TODO اصلی بنویسید تا بتوان هر تمرین را جدا آزمود.</p>
</div>

```python
rules = [("a", "b"), ("ab", "c")]
wrong_order = list(reversed(rules))
pieces = list("abc")
for left, right in wrong_order:
    index = 0
    while index+1 < len(pieces):
        if pieces[index:index+2] == [left, right]:
            pieces[index:index+2] = [left+right]
        index += 1
print("Rules in reverse order:", pieces)
assert pieces == ["ab", "c"]
```
