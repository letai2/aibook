<div dir="rtl">
<h2>فقط یک عامل را تغییر دهید</h2>
<p>فقط اندازهٔ Batch را از یک به دو تغییر دهید؛ وزن ثابت و Seed ثابت بمانند. چند Gradient را ثبت کنید، نه فقط آخرین عدد را.</p>
</div>

```python
for size in [1, 2]:
    rng = random.Random(42)
    estimates = []
    for _ in range(8):
        selected = rng.choices(data, k=size)
        estimates.append(sum(-2*y*x for x, y in selected)/size)
    print("batch size:", size, "gradients at w=0:", estimates)
```
