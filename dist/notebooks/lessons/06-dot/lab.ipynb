<h2 style="text-align:right">فقط یک عامل را تغییر دهید</h2>
<p style="text-align:right">فقط اندازهٔ <code dir="ltr" style="direction:ltr;text-align:left;unicode-bidi:isolate">b</code> را دو برابر کنید، نه جهت آن را. سپس فقط وزن ترجیح اول را تغییر دهید. در هر آزمایش بگویید چه چیزی ثابت مانده است.</p>

```python
for candidate in [b, [2*value for value in b]]:
    print("candidate:", candidate, "score:", sum(x*y for x, y in zip(q, candidate)))
print("Changed preference:", sum(x*y for x, y in zip([1, 1], b)))
```
