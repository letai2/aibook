<h2>فقط یک عامل را تغییر دهید</h2>
<p>فقط اندازهٔ b را دو برابر کنید، نه جهت آن را. سپس فقط وزن ترجیح اول را تغییر دهید. در هر آزمایش بگویید چه چیزی ثابت مانده است.</p>
</div>

```python
for candidate in [b, [2*value for value in b]]:
    print("candidate:", candidate, "score:", sum(x*y for x, y in zip(q, candidate)))
print("Changed preference:", sum(x*y for x, y in zip([1, 1], b)))
```
